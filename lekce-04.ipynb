# Pivot tabulky

```python
import pandas as pd
```

```python
pd.options.display.float_format = '{:.2f}'.format
# pd.reset_option('display.float_format')
```

```python
cholesterol = pd.read_csv("pivot_cholesterol.csv")
cholesterol.head()
```

```python
# 0-20 nizky
# 20-100 stredni
# vice nez 100 vysoky
# bins & labels - standardni oznaceni

bins = [0, 20, 100, float('inf')]
labels = ["nizky", "stredni", "vysoky"]

## funkce CUT
cholesterol["category"] = pd.cut(cholesterol["amount"], bins=bins, labels=labels)
cholesterol
```

```python
pd.crosstab(cholesterol["branded_food_category"], cholesterol["category"])
```

```python
# funkce CROSSTAB

pd.crosstab(index=cholesterol["branded_food_category"], 
            columns=cholesterol["category"],
            margins=True,
            normalize="columns").style.format("{:.2%}")
```

```python
cholesterol_fiber = pd.read_csv("pivot_cholesterol_fiber.csv")
cholesterol_fiber.head()
```

```python
pd.crosstab(index=cholesterol_fiber["branded_food_category"],
            columns=cholesterol_fiber["nutrient_name"])
```

```python
pd.crosstab(index=cholesterol_fiber["branded_food_category"],
            columns=cholesterol_fiber["nutrient_name"],
            values=cholesterol_fiber["amount"],
            aggfunc="mean")
```

```python
pd.crosstab(hotel_bookings["hotel"], hotel_bookings["is_canceled"], margins=True)
```

```python
pd.crosstab(city["Reservation_Category"], city["is_canceled"], margins=True)
```

# Datum a čas a metoda shift

```python
signal_monitoring = pd.read_csv("signal_monitoring.csv")
signal_monitoring
```

```python
signal_monitoring["event_date_time"] = pd.to_datetime(signal_monitoring["event_date_time"])
```

```python
signal_monitoring["event_date"] = signal_monitoring["event_date_time"].dt.date
```

```python
signal_monitoring["event_date_time"].dt.daysinmonth
```

```python
signal_monitoring["event_end_date_time"] = signal_monitoring["event_date_time"].shift(periods=-1)
signal_monitoring
```

```python
signal_monitoring = signal_monitoring[signal_monitoring["event_type"] == "signal lost"].reset_index(drop=True)
signal_monitoring
```

```python
signal_monitoring["outage_length"] = signal_monitoring["event_end_date_time"] - signal_monitoring["event_date_time"]
signal_monitoring
```

```python
signal_monitoring["event_date"].value_counts()
```

```python
signal_monitoring.groupby("event_date")["outage_length"].sum()
```

## Cvičení

### Půlmaraton

```python
half_maraton = pd.read_csv("half_marathon.csv")
half_maraton.head()
```

```python
half_maraton["Cas"] = pd.to_datetime(half_maraton["Cas"])
```

```python
half_maraton["Cas_2019"] = half_maraton["Cas"].shift(1)
```

```python
half_maraton = half_maraton[half_maraton["Rok zavodu"] == 2020]
```

```python
half_maraton["Rozdil_v_case"] = (half_maraton["Cas"] - half_maraton["Cas_2019"]).dt.total_seconds()
```

```python
print("Průměrná změna v sekundách: ", half_maraton["Rozdil_v_case"].mean())
print("Průměrná změna v minutách: ", half_maraton["Rozdil_v_case"].mean() /60)
print("Průměrná absolutní změna v minutách: ", half_maraton["Rozdil_v_case"].abs().mean() /60)
```

### Swing states

```python
data = pd.read_csv("election-data.csv")
data = data[data["rank"] == 1]
```

```python
data_pen = data[data["state"] == "PENNSYLVANIA"]
data_pen["party_simplified_previous_election"] = data_pen["party_simplified"].shift()
data_pen = data_pen.dropna()
data_pen["swing"] = data_pen["party_simplified"] != data_pen["party_simplified_previous_election"]
data_pen[data_pen["swing"] == 1]
```

```python
data_pen["swing"].sum()
```

```python
data["party_simplified_previous_election"] = data.groupby("state")["party_simplified"].shift(1)
data = data.dropna()
data["swing"] = data["party_simplified"] != data["party_simplified_previous_election"]
data.groupby("state")["swing"].sum().sort_values(ascending=False)
```
