```python
average_mass = data.groupby('species')['body_mass_g'].mean()
print(average_mass)
```

```python
max_mass = data.groupby('species')['body_mass_g'].max()
```

```python
max_mass_sorted = max_mass.sort_values(ascending=False)
max_species = max_mass_sorted.index[0]
max_value = max_mass_sorted.iloc[0]
```

```python
print(f"Druh s největší hmotností: {max_species}")
```

```python
print(f"Maximální hmotnost: {max_value} g")
```

```python
sex_counts = data.groupby(['species', 'sex']).size()
print(sex_counts)
```

```python
flipper_stats = data.groupby('species')['flipper_length_mm'].agg(['min', 'max'])
```

```python
flipper_stats['range'] = flipper_stats['max'] - flipper_stats['min']
print(flipper_stats)
```

```python
min_range_species = flipper_stats['range'].idxmin()
min_range_value = flipper_stats['range'].min()
```

```python
print(f"\nDruh s nejmenším rozsahem délky křídel: {min_range_species}")
print(f"Rozdíl mezi max a min: {min_range_value} mm")
```

```python
median_mass_by_island = data.groupby('island')['body_mass_g'].median()
print("Medián hmotnosti tučňáků podle ostrova:")
print(median_mass_by_island)
```

```python
max_median_island = median_mass_by_island.idxmax()
max_median_value = median_mass_by_island.max()
print(f"\nOstrov s největším mediánem hmotnosti: {max_median_island}")
print(f"Medián hmotnosti na tomto ostrově: {max_median_value} g")
```

```python
average_flipper = data.groupby('species')['flipper_length_mm'].mean()
```

```python
species_over_200 = average_flipper[average_flipper > 200]
print("Druhy tučňáků s průměrnou délkou křídel větší než 200 mm:")
print(species_over_200)
```

## 4. úkol

```python
bins = [0, 180, 200, float("inf")]
labels = ['short', 'medium', 'long']
```

```python
data['flipper_category'] = pd.cut(
    data['flipper_length_mm'],
    bins=bins,
    labels=labels,
    #include_lowest=True
)


print(data[['flipper_length_mm', 'flipper_category']].head())
print(data['flipper_category'].value_counts())
```

```python
data["flipper_category"].value_counts().get("long", 0)
```

```python
pd.crosstab(data['species'], data['island'])
```

```python
pd.crosstab(data["species"], data["island"], normalize="index")
```

```python
pivot = data.pivot_table(
    values='bill_depth_mm',  # co agregovat
    index='species',         # řádky = druh tučňáka
    columns='sex',           # sloupce = pohlaví
    aggfunc='mean'           # agregační funkce = průměr
)

print(pivot)
```

```python
# Pivot tabulka: průměrná hloubka zobáku podle druhu a pohlaví
pivot = data.pivot_table(
    values='bill_depth_mm',  # co agregujeme
    index='species',         # řádky = druh tučňáka
    columns='sex',           # sloupce = pohlaví
    aggfunc='mean'           # agregační funkce = průměr
)

print(pivot)

# Druh s nejkratší průměrnou hloubkou zobáku u samic
druh_nejkratší_zobak = pivot['female'].idxmin()
print("Samičky druhu s nejkratší hloubkou zobáku:", druh_nejkratší_zobak)
```
